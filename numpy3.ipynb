## 📌 Indexing and Slicing in NumPy

### 🔹 Indexing

Indexing in NumPy works similarly to Python lists. You can access individual elements of a NumPy array using their index positions. Positive indices access from the beginning, while negative indices count from the end of the array.

### 🔹 Slicing

Slicing refers to extracting a portion or subset of an array using a start\:stop\:step syntax. It allows you to extract specific ranges or patterns of elements.

* **Start**: The index to begin the slice (inclusive).
* **Stop**: The index to stop (exclusive).
* **Step**: The interval between elements.

### ⚠️ Important: Slicing Returns a View, Not a Copy

Unlike Python lists, slicing a NumPy array does **not** create a new array—it returns a **view** of the original array. This means that changes made to the sliced array will directly reflect in the original array.

#### Why does slicing return a view in NumPy?

* **Memory Efficiency**: Prevents unnecessary duplication of data, which is especially beneficial for large datasets.
* **Performance**: Enables fast manipulation and access without allocating additional memory.

If an independent copy is required, the `.copy()` method should be used to avoid altering the original data.

## 🔸 Fancy Indexing & Boolean Masking

### ✅ Fancy Indexing

Fancy indexing lets you access multiple elements of an array at once by passing a list or array of specific indices. This is useful when you need non-contiguous or specifically selected elements.

### ✅ Boolean Masking

Boolean masking is a powerful feature that allows filtering of data using boolean conditions. A boolean array (mask) is created by applying a condition, and this mask is used to extract elements that meet the condition.

Boolean masking is highly efficient and is commonly used for filtering and conditional data operations in large datasets.

---

## 📝 Summary

* NumPy arrays provide a faster, memory-efficient alternative to Python lists.
* Arrays can be created using various functions like `np.array()`, `np.zeros()`, `np.ones()`, and more.
* Indexing and slicing allow direct and flexible access to array data.
* Slicing returns a **view**, not a **copy**, which shares memory with the original array.
* Use `.copy()` to create a true duplicate of an array slice.
* Fancy indexing and boolean masking provide efficient ways to filter and access data based on custom conditions.
* These operations make NumPy a powerful tool for handling large numerical datasets efficiently.

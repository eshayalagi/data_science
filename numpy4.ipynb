**Multidimensional Indexing and Axis**

Here’s a **theory-based explanation** of *Multidimensional Indexing and Axis* in NumPy, without any code:

---

### **1. Understanding Axes in NumPy**

In NumPy, an array can have one or more dimensions. Each dimension is called an **axis**, and they are numbered starting from **0**.

* A **1D array** has only **one axis** (axis 0), representing elements in a single line.
* A **2D array** has **two axes**:

  * **Axis 0** runs *vertically* (along the rows).
  * **Axis 1** runs *horizontally* (along the columns).
* A **3D array** introduces a third axis:

  * **Axis 0** refers to the *depth* or number of 2D "sheets".
  * **Axis 1** refers to the *rows* within each sheet.
  * **Axis 2** refers to the *columns* within each sheet.

Understanding these axes is critical for performing operations like summing, reshaping, and transforming arrays.

---

### **2. Indexing in Multidimensional Arrays**

**Indexing** is the method used to access specific elements or groups of elements within an array.

* In a **2D array**, the index format is `[row, column]`.
* In a **3D array**, it is `[depth, row, column]`.

You can extract single elements or entire sections using **slicing**, which allows you to specify a range of indices for any axis. This helps in working with parts of large arrays without copying or looping manually.

---

### **3. Axis-Based Operations**

NumPy enables efficient operations along specific axes:

* When you perform an operation **along axis 0**, it means the operation is applied **down each column**, across all rows.
* When you perform an operation **along axis 1**, it applies **across each row**, moving through columns.

These operations return results reduced across the specified axis, enabling fast summarization or transformation of data.

---

### **4. Working with 3D Arrays**

In 3D arrays:

* You work with multiple 2D arrays stacked along the first axis (depth).
* You can access a specific "sheet", or slice across all sheets.
* For instance, selecting all first rows from every depth level lets you analyze consistent positions across layers of data.

This is commonly used in fields like image processing (color channels), scientific simulations (time steps), or deep learning (batch data).

---

### **5. Modifying Data Along an Axis**

NumPy lets you directly alter elements along any axis. For example:

* You can update all values in a column or row at once.
* This is done using slicing or indexing and is far more efficient than using loops.

Modifying values along a specific axis allows batch updates, which is vital in data preprocessing or feature engineering.

---

### **6. Summary of Key Concepts**

* **Axes** represent dimensions; axis numbering starts at 0.
* **Axis 0**: vertical (rows), **Axis 1**: horizontal (columns), **Axis 2**: depth (for 3D).
* **Indexing and slicing** help access and manipulate specific parts of arrays.
* **Operations along axes** (like sum, mean, etc.) reduce or transform data effectively.
* **Modifying data** along an axis helps with structured updates without loops.

Let’s **break down how to understand and work with 3D arrays** in NumPy, conceptually—without code.

---

### 🔷 What is a 3D Array?

A **3D array** in NumPy is like a collection of 2D arrays stacked together.

You can imagine it as:

* A **book of tables** (each page is a 2D table),
* Each **page** is a matrix (with rows and columns),
* The **book** is the entire 3D array.

So, it has:

* **Axis 0 (depth):** selects which 2D array (which "page" of the book),
* **Axis 1 (rows):** selects which row in that 2D array,
* **Axis 2 (columns):** selects which column in that row.

---

### 🔷 Shape of a 3D Array

The shape is written as:
`(number_of_2D_arrays, number_of_rows, number_of_columns)`

Example:
`(2, 2, 3)` means:

* 2 "sheets" (pages),
* Each sheet has 2 rows and 3 columns.

Visually:

```
Sheet 0:
[[1, 2, 3],
 [4, 5, 6]]

Sheet 1:
[[7, 8, 9],
 [10,11,12]]
```

---

### 🔷 How to Access Data in 3D Arrays

You access elements using **three indices**:
`[depth_index, row_index, column_index]`

* `array[0, 1, 2]` → from **sheet 0**, **row 1**, **column 2**
* `array[1, 0, :]` → from **sheet 1**, **row 0**, **all columns**
* `array[:, 0, :]` → from **all sheets**, **row 0**, **all columns**
  → gives you the first row from every sheet

---

### 🔷 Common Uses of 3D Arrays

* **Images:** (height, width, color channels)
* **Video frames:** (frame number, height, width)
* **Deep Learning:** (batch size, features, time steps)

---

### 🔷 Summary

| Dimension | Meaning             | Axis # |
| --------- | ------------------- | ------ |
| Sheets    | Depth (pages)       | 0      |
| Rows      | Rows of each sheet  | 1      |
| Columns   | Columns of each row | 2      |

Understanding this lets you extract, modify, or compute data across layers, rows, or columns in large datasets easily.

---

## ✅ **2D Array (Books Table)**

### 📚 Real-life Analogy:

A **2D array** is like a **single table of books**, where each row is a book and each column is a detail (like ID, price, quantity).

```python
books = np.array([
    [101, 250, 3],   # Book 1
    [102, 300, 5],   # Book 2
    [103, 150, 2]    # Book 3
])
```

### 🔹 Shape: `(3, 3)`

* 3 rows → 3 books
* 3 columns → ID, Price, Quantity

### 🔹 Access:

* `books[1, 2]` → Row 1, Column 2 → Book 2’s quantity → `5`

---

## ✅ **3D Array (Multiple Book Tables)**

### 📘 Real-life Analogy:

A **3D array** is like having **multiple book tables** (like for different branches, stores, or years).
Each "sheet" is a 2D book table.

```python
books3D = np.array([
    [  # Branch 1
        [101, 250, 3],
        [102, 300, 5]
    ],
    [  # Branch 2
        [201, 400, 6],
        [202, 350, 2]
    ]
])
```

### 🔹 Shape: `(2, 2, 3)`

* 2 tables (branches)
* 2 books per branch
* 3 details per book (ID, Price, Quantity)

### 🔹 Access:

* `books3D[1, 0, 1]`
  → Branch 2, Book 1, Price → `400`

* `books3D[:, 0, :]`
  → First book from **all branches**
  → Output:

  ```
  [[101, 250, 3],
   [201, 400, 6]]
  ```

---

## 🔄 Comparison Table

| Feature    | 2D Array                | 3D Array                                  |
| ---------- | ----------------------- | ----------------------------------------- |
| Structure  | Table (rows × columns)  | Stack of tables (sheets × rows × columns) |
| Real-world | Book list for one store | Book lists for multiple stores            |
| Shape      | `(rows, columns)`       | `(sheets, rows, columns)`                 |
| Access     | `array[row, column]`    | `array[sheet, row, column]`               |
| Use Case   | Flat structured data    | Time-series, branches, channels, datasets |

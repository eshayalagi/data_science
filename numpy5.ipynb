## 📌 NumPy Data Types – Theory

### 1. Introduction to NumPy Data Types

NumPy arrays are **homogeneous**, meaning all elements must be of the same data type. This is unlike Python lists, which can contain mixed types. Choosing the right data type (`dtype`) in NumPy is essential for efficient memory usage and performance in data-heavy applications.

### 2. Common NumPy Data Types

* **Integer Types**: `int32`, `int64` – Used for whole numbers, differing in memory size.
* **Float Types**: `float32`, `float64` – Used for decimal numbers with varying levels of precision.
* **Boolean**: `bool` – Stores `True` or `False` values, very memory-efficient.
* **Complex Numbers**: `complex64`, `complex128` – Stores numbers with real and imaginary parts.
* **String**: `str` or `U` – Stores fixed-size Unicode strings.
* **Object**: `object` – Allows mixed types like dictionaries, lists, or strings. Less efficient.

### 3. Changing Data Types

NumPy allows you to **cast arrays to another type** using `.astype()`. This is often done to:

* Match the required precision
* Save memory (e.g., casting `int64` to `int32`)
* Perform type-specific operations

### 4. Memory and Performance Implications

The data type of a NumPy array directly impacts:

* **Memory Usage**: Smaller data types occupy less memory.
* **Speed**: Operations on smaller data types are generally faster.
* **Accuracy**: Using a float type when needed avoids truncation errors.

### 5. String and Object Arrays

Although NumPy supports strings and Python objects, they are:

* **Less efficient** compared to numeric types
* **Slower** for operations like searching or sorting
* **Memory-intensive**, especially when using `dtype=object`

Use them only when absolutely necessary.

### 6. Complex Numbers Support

NumPy natively supports complex numbers with real and imaginary parts. These are stored using specific types like `complex64` and `complex128`.

### 7. Best Practices

* Always choose the **smallest possible data type** that meets your needs.
* Use **numeric types** (int, float) for speed and memory efficiency.
* Avoid `object` arrays unless you're dealing with truly mixed data types.
* Downcast data types to reduce memory usage in large datasets.

```python
import numpy as np
arr=np.array([1])
print(arr.dtype)
```

```python
arr = np.array([1.5, 2.7, 3.9])
print(arr.dtype) 
```

```python
arr_int = arr.astype(np.int32)  # Converting float to int
print(arr_int)    # Output: [1 2 3]
print(arr_int.dtype) 
```

```python
arr_large = np.array([1000000, 2000000, 3000000], dtype=np.int64)
arr_small = arr_large.astype(np.int32)  # Downcasting to a smaller dtype
print(arr_small)  # Output: [1000000 2000000 3000000]
print(arr_small.dtype)  # Output: int32
```

You're correctly using `.nbytes` to check the memory consumption of NumPy arrays! Since `int64` uses 8 bytes per element and `int32` uses 4 bytes per element, your results are expected:

- **`arr_int64.nbytes`** → `3 * 8 = 24 bytes`
- **`arr_int32.nbytes`** → `3 * 4 = 12 bytes`

If you're working with large datasets, choosing the right dtype can help optimize memory usage. Do you want to explore further memory profiling techniques? 🚀


```python
arr_int64 = np.array([1, 2, 3], dtype=np.int64)
arr_int32 = np.array([1, 2, 3], dtype=np.int32)

print(arr_int64.nbytes)  # Output: 24 bytes (3 elements * 8 bytes each)
print(arr_int32.nbytes)  # Output: 12 bytes (3 elements * 4 bytes each)
```

```python
import numpy as np

# 1. Creating a Unicode string array with a different fixed size
arr1 = np.array(['pineapple', 'kiwi', 'grape'], dtype='U8')
print(arr1)  # Max character length is 8

# 2. Truncation example
arr2 = np.array(['strawberry', 'blueberry', 'raspberry'], dtype='U6')
print(arr2)  # Strings are truncated to 6 characters

# 3. Combining different string lengths
arr3 = np.array(['apple', 'banana', 'mango', 'watermelon'], dtype='U12')
print(arr3)  # Max character length is 12

# 4. Modifying the dtype dynamically
arr4 = np.array(['hello', 'world'], dtype='U5')
arr4 = arr4.astype('U10')  # Changing the dtype to allow longer strings
print(arr4)

# 5. Sorting Unicode arrays
arr5 = np.array(['orange', 'pear', 'apple', 'banana'], dtype='U10')
sorted_arr = np.sort(arr5)
print("Sorted array:", sorted_arr)
```

```python
arr = np.array([1 + 2j, 3 + 4j, 5 + 6j], dtype='complex128')
print(arr)
```

```python
import numpy as np

arr = np.array([1 + 2j, 3 + 4j, 5 + 6j], dtype='complex128')
print(arr)
```

```python
print("Real parts:", arr.real)  # [1. 3. 5.]
print("Imaginary parts:", arr.imag)  # [2. 4. 6.]
print("Absolute values:", np.abs(arr))  # [2.236 5. 7.81] (Magnitude of complex numbers)
```

Integer (int)
np.int8 → 1 byte (8 bits) → Range: -128 to 127

np.int16 → 2 bytes (16 bits) → Range: -32,768 to 32,767

np.int32 → 4 bytes (32 bits) → Range: -2,147,483,648 to 2,147,483,647

np.int64 → 8 bytes (64 bits) → Range: -9.2 quintillion to 9.2 quintillion

Float (float)
np.float16 → 2 bytes → Lower precision, limited range

np.float32 → 4 bytes → Standard single-precision float

np.float64 → 8 bytes → Higher precision, used by default
